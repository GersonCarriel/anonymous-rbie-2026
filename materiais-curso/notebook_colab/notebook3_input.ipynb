# Notebook 3 – Entrada de dados com `input()`

Neste notebook, você vai aprender a **entrada de dados**, ou seja, como o usuário pode digitar informações
que o programa vai usar nos cálculos.

Vamos ligar isso ao modelo:

**Entrada → Processamento → Memória → Saída**.

Aqui o foco principal será a **Entrada**, usando o comando `input()`.

## 1. Primeiro exemplo de `input()`

O comando `input()` mostra uma pergunta na tela e espera o usuário digitar algo.

Execute o código abaixo, digite seu nome quando for pedido e pressione **Enter**.

```
nome = input("Digite seu nome: ")
print("Olá,", nome, "! Bem-vindo(a) à aula de Python.")
```

## 2. Digitando números (sem conversão ainda)

Agora vamos pedir dois números ao usuário. **Neste momento, não vamos fazer nenhuma conversão**,
vamos apenas somar os valores como o Python entende por padrão.

Tente, por exemplo, digitar 2 e 3.

```
a = input("Digite um número: ")
b = input("Digite outro número: ")
c = a + b
print("Resultado da soma:", c)
```

## 3. O que aconteceu com a "soma"?

Repare no resultado. Se você digitou 2 e 3, provavelmente apareceu `23` na tela.

Isso acontece porque, do jeito que está, o Python está tratando aquilo que você digitou como **texto**, e não como número.

Por enquanto, não vamos resolver esse problema. A ideia é você **perceber** que algo diferente do esperado aconteceu.

Na próxima aula, vamos entender como **converter** esses valores para números (inteiros e reais).

## 4. Outro exemplo com texto

Também podemos usar `input()` para outros tipos de perguntas. Execute o código abaixo
e digite as respostas que fizerem sentido para você.

```
curso = input("Qual curso você está fazendo? ") 
cidade = input("De qual cidade você está acessando esta aula? ") 
print("Você está no curso", curso, "e acessando a aula a partir de", cidade + ".")
```

## 6. Encerrando

Neste notebook, você viu que:

- `input()` permite que o usuário **digite dados** para o programa;
- O que é digitado entra como **texto** (string) por padrão;
- Quando tentamos somar textos que parecem números, o resultado pode ficar estranho (como `2` + `3` → `23`);
