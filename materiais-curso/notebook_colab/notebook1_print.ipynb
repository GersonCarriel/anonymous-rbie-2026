# Notebook 1 – Primeiros passos com `print()`

Neste notebook, vamos treinar a **saída de dados na tela** usando o comando `print()` em Python.

Você só precisa **executar as células** (botão ▶) e observar o resultado.

## 1. Seu primeiro `print`

O comando `print()` serve para **mostrar informações na tela**.

Clique no botão ▶ ao lado da célula abaixo e veja o que acontece.

```
print("Olá, mundo!")
```

## 2. Mostrando textos e números

Podemos mostrar **textos** (entre aspas) e **números** (sem aspas).

Execute os exemplos a seguir.

```
print("Este é um texto.")
print(10)
print(3.14)
print("Hoje é aula de Python!")
```

## 3. Combinando texto com resultados

Também podemos mostrar um texto junto com o resultado de um cálculo.

Repare que o cálculo é feito **dentro** do `print()`, e o resultado aparece na tela.

```
print("A soma de 2 + 3 é:", 2 + 3)
print("O dobro de 10 é:", 10 * 2)
print("A média entre 8 e 6 é:", (8 + 6) / 2)
```

## 4. Pequena atividade

Agora é sua vez:

1. Altere o texto do primeiro `print()` abaixo para mostrar seu nome.
2. Altere os números do segundo `print()` para fazer outra conta.
3. Depois clique em ▶ para ver o resultado.

Não se preocupe se errar, você pode ajustar e executar novamente.

```
print("Escreva aqui uma mensagem com seu nome")
print("O resultado de 10 + 5 é:", 10 + 5)
```

## 5. Encerrando

Neste notebook, você viu que:

- `print()` é o comando usado para **mostrar informações na tela**;
- Podemos mostrar **textos**, **números** e **resultados de contas**;
- É possível combinar texto e cálculo em um mesmo `print()`.
