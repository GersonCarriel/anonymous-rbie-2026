# Notebook 2 – Variáveis e Processamento em Python

Neste notebook, vamos ver como o computador **guarda valores na memória** usando variáveis,
e como realiza **processamentos** (contas) com esses valores.

Lembre-se do modelo: **Entrada → Processamento → Memória → Saída**.
Aqui vamos focar em **Memória + Processamento**.

## 1. O que é uma variável?

Uma variável é um **nome** que usamos para guardar um valor na memória.

Por exemplo, no código abaixo, `a` e `b` são variáveis que armazenam números.

```
a = 10
b = 5
print(a)
print(b)
```

## 2. Fazendo contas com variáveis

Podemos usar as variáveis em expressões matemáticas. O computador usa os valores guardados nelas
para fazer o cálculo.

Execute o exemplo a seguir:

```
a = 10
b = 5
c = a + b
print("Valor de a:", a)
print("Valor de b:", b)
print("Soma (c = a + b):", c)
```

## 3. Outros exemplos de processamento

Aqui estão mais alguns exemplos de contas usando variáveis:
- soma
- subtração
- multiplicação
- divisão

Execute e observe os resultados.

```
x = 8
y = 3
soma = x + y
subtracao = x - y
multiplicacao = x * y
divisao = x / y
print("x =", x)
print("y =", y)
print("Soma:", soma)
print("Subtração:", subtracao)
print("Multiplicação:", multiplicacao)
print("Divisão:", divisao)
```
